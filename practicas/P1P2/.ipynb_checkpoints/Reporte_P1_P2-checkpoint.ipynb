### 1. Definición de URDF
**URDF** son las siglas de *Unified Robot Description Format* (Formato Unificado de Descripción de Robots). Es un formato de archivo basado en **XML** utilizado en **ROS** (*Robot Operating System*) para describir todos los elementos físicos de un robot. Este archivo define la estructura cinemática y dinámica del robot, incluyendo:
* Sus eslabones (*links*).
* Las articulaciones (*joints*) que los unen.
* Sus propiedades visuales, de colisión y de inercia.

### 2. Tipos de juntas definidas dentro del formato URDF
El formato URDF especifica varios tipos de articulaciones (*joints*) que definen cómo se mueve un eslabón respecto a otro. Los tipos principales son:

* **Revolute (Revoluta):** Una articulación de bisagra que rota a lo largo de un eje especificado y tiene límites de rango superior e inferior (limitada).
* **Continuous (Continua):** Similar a la revoluta, pero sin límites de rotación (puede girar infinitamente, como una rueda).
* **Prismatic (Prismática):** Una articulación deslizante que se mueve linealmente a lo largo de un eje y tiene límites de rango.
* **Fixed (Fija):** No permite ningún movimiento; une dos eslabones de manera rígida (como si fueran uno solo).
* **Floating (Flotante):** Permite movimiento en 6 grados de libertad (traslación y rotación en los 3 ejes).
* **Planar (Planar):** Permite movimiento en un plano perpendicular a un eje.

### 3. Definición de elementos "primitivos" en ROS
Para definir la geometría visual o de colisión de un robot en un URDF sin necesidad de importar mallas complejas (archivos .stl o .dae), se utilizan figuras geométricas básicas conocidas como primitivas:

* **Box (Caja):** Definida por sus dimensiones en los ejes x, y, z (`size`).
* **Cylinder (Cilindro):** Definido por su radio (`radius`) y longitud (`length`).
* **Sphere (Esfera):** Definida únicamente por su radio (`radius`).

Estas primitivas son computacionalmente más ligeras y son ideales para definir las propiedades de colisión (`<collision>`) del robot.

### 4. Nodo y Tópico en ROS 2
* **Nodo (Node):** Es la unidad fundamental de computación en ROS 2. Un nodo es un proceso ejecutable que realiza una tarea específica (ej. leer un sensor, controlar un motor, procesar una imagen). Un sistema robótico suele estar compuesto por muchos nodos que trabajan en conjunto.
* **Tópico (Topic):** Es el mecanismo de comunicación mediante el cual los nodos intercambian información. Funciona bajo un modelo de publicación/suscripción. Un nodo "publica" datos en un tópico y cualquier otro nodo que necesite esa información se "suscribe" a dicho tópico para recibirla en tiempo real.

### 5. Qué es RVIZ
**RVIZ (ROS Visualization)** es una herramienta de visualización 3D para ROS. A diferencia de un simulador (como Gazebo), RVIZ no calcula física; su función es mostrar lo que el robot "ve" y "cree". Permite visualizar:
* El modelo del robot (basado en el URDF).
* Datos de sensores en tiempo real (Lidar, cámaras, nubes de puntos).
* Sistemas de coordenadas (TF).
* Mapas y trayectorias planificadas.

### 1. El modelo de un robot móvil diferencial (2,0)
El robot móvil diferencial, denotado como tipo (2,0), es una configuración clásica en robótica que consta de dos ruedas motrices fijas alineadas sobre un eje común y, generalmente, una o más ruedas locas (*caster wheels*) para mantener el equilibrio estático.

* **Cinemática:** El movimiento del robot se describe mediante la relación entre las velocidades de sus ruedas (izquierda y derecha) y su velocidad lineal y angular en el plano.
* **Restricciones:** Este robot posee restricciones no holonómicas, lo que significa que no puede desplazarse instantáneamente en la dirección del eje de las ruedas (no puede deslizarse lateralmente).
* **Ecuaciones:** La velocidad lineal ($v$) y angular ($\omega$) del robot se calculan como:
    $$v = \frac{R}{2} (\dot{\phi}_R + \dot{\phi}_L)$$
    $$\omega = \frac{R}{L} (\dot{\phi}_R - \dot{\phi}_L)$$
    Donde $R$ es el radio de las ruedas, $L$ la distancia entre ellas, y $\dot{\phi}$ la velocidad angular de cada rueda.



### 2. Principios de funcionamiento de un sensor LIDAR
El sensor **LIDAR** (*Light Detection and Ranging*) es un dispositivo activo utilizado para medir distancias y generar mapas del entorno con alta precisión.

* **Principio de Tiempo de Vuelo (ToF):** El sensor emite pulsos de luz láser (infrarroja) que rebotan en los objetos y regresan al detector. La distancia ($d$) se calcula midiendo el tiempo ($t$) que tarda el pulso en ir y volver:
    $$d = \frac{c \cdot t}{2}$$
    (Donde $c$ es la velocidad de la luz).
* **Mapeo 2D:** En robots móviles como los usados en clase (Turtlebot), el láser gira o utiliza espejos rotatorios para escanear un plano de 360°, generando una nube de puntos bidimensional que representa los obstáculos a su alrededor.



### 3. Principio de funcionamiento del algoritmo de navegación SLAM
**SLAM** (*Simultaneous Localization and Mapping*) es un algoritmo fundamental en robótica móvil que resuelve el problema de "el huevo y la gallina": el robot necesita un mapa para saber dónde está, pero necesita saber dónde está para construir el mapa.

* **Funcionamiento:**
    1.  **Predicción:** El robot estima su movimiento basándose en la odometría (datos de las ruedas).
    2.  **Percepción:** El robot utiliza sus sensores (como el LIDAR) para detectar características del entorno.
    3.  **Corrección:** El algoritmo compara lo que "ve" el sensor con lo que esperaba ver según su mapa provisional. Utiliza métodos probabilísticos (como Filtros de Kalman o Filtros de Partículas) para corregir tanto la estimación de su posición como la estructura del mapa.

### 4. Implementación del mapeo de un entorno utilizando las herramientas de RVIZ2
En ROS 2, la visualización del proceso de mapeo se realiza configurando **RVIZ2** para suscribirse a los tópicos generados por el nodo de SLAM (como `slam_toolbox` o `cartographer`).

* **Tópico del Mapa:** Se debe añadir una visualización del tipo **Map** suscrita al tópico `/map`. Esto muestra una rejilla de ocupación (*OccupancyGrid*) donde:
    * **Celdas Blancas:** Espacio libre.
    * **Celdas Negras:** Obstáculos/Muros.
    * **Celdas Grises:** Espacio desconocido.
* **Datos del Sensor:** Se añade una visualización del tipo **LaserScan** suscrita al tópico `/scan` para ver en tiempo real qué está detectando el robot.
* **Sistemas de Referencia:** Es crucial configurar el *Fixed Frame* en `map` u `odom` para que los datos visuales sean coherentes con el movimiento del robot.

### 5. Qué es RVIZ
**RVIZ** (*ROS Visualization*) es la herramienta gráfica estándar en ROS para visualizar el estado del robot y los datos de sus sensores.

* **Función:** A diferencia de un simulador (como Gazebo), RVIZ no calcula física ni colisiones. Su propósito es mostrar al usuario lo que el robot "piensa" y "ve".
* **Utilidad:** Permite visualizar modelos URDF, nubes de puntos, imágenes de cámaras, trayectorias planificadas y, fundamentalmente, el mapa generado por algoritmos de SLAM.

# **Sintonización del Controlador (PID)**

Para lograr un control de posición preciso, fue necesario configurar los controladores de cada articulación. Inicialmente, el robot presentaba oscilaciones severas debido a la falta de amortiguamiento en la simulación física.

### Estrategia de Control
Se implementó un esquema **PID (Proporcional-Integral-Derivativo)** en el archivo de descripción URDF (`scara_gz2.xacro`) con los siguientes criterios:

* **Alta Rigidez ($K_p$):** Se incrementó la ganancia proporcional para asegurar que el robot tuviera la fuerza necesaria para moverse rápidamente.
* **Amortiguamiento Crítico ($K_d$):** Se añadieron ganancias derivativas altas. Dado que Gazebo simula un entorno ideal (sin fricción mecánica), fue necesario un valor alto de $D$ para disipar la energía cinética y lograr un frenado "seco" sin rebotes.
* **Corrección de Error ($K_i$):** Se añadió una pequeña ganancia integral con límites (`clamping`) para corregir errores de estado estacionario sin causar inestabilidad (*anti-windup*).

### Mecanismos de Seguridad
* **Anti-Windup:** Es una estrategia de protección para el término Integral. Si el robot se bloquea mecánicamente, la parte Integral ($I$) tendería a acumular fuerza infinitamente hasta niveles peligrosos. El *Anti-Windup* es la regla que dicta: "Deja de acumular fuerza si el actuador ya llegó a su límite".
* **Clamping (Saturación):** Es la técnica específica implementada para lograr el *Anti-Windup*. Consiste en establecer un "techo" y un "piso" numérico (límites máximos y mínimos) al valor de la Integral en el código, recortando cualquier valor que exceda ese rango seguro.


A continuación se presenta el código final implementado en la descripción del robot. Este bloque define los parámetros de control sintonizados para eliminar oscilaciones y garantizar precisión.


````python
<?xml version="1.0"?>
<robot xmlns:xacro="http://www.ros.org/wiki/xacro">

    <gazebo>
        <plugin 
            filename="gz-sim-joint-position-controller-system" 
            name="gz::sim::systems::JointPositionController">
            <joint_name>link_1_joint</joint_name>
            
            <p_gain>200.0</p_gain>
            
            <d_gain>65.0</d_gain>
            
            <i_gain>0.01</i_gain>
            <i_max>1.0</i_max>   <i_min>-1.0</i_min>  
        </plugin>
    </gazebo>

    <gazebo>
        <plugin 
            filename="gz-sim-joint-position-controller-system" 
            name="gz::sim::systems::JointPositionController">
            <joint_name>link_2_joint</joint_name>
            
            <p_gain>100.0</p_gain>
            <d_gain>30.0</d_gain>
            
            <i_gain>0.01</i_gain>
            <i_max>1.0</i_max>
            <i_min>-1.0</i_min>
        </plugin>
    </gazebo>

    <gazebo>
        <plugin 
            filename="gz-sim-joint-position-controller-system" 
            name="gz::sim::systems::JointPositionController">
            <joint_name>link_3_joint</joint_name>
            
            <p_gain>100.0</p_gain>
            <d_gain>30.0</d_gain>
            
            <i_gain>0.01</i_gain>
            <i_max>1.0</i_max>
            <i_min>-1.0</i_min>
        </plugin>
    </gazebo>

    <gazebo>
        <plugin 
            filename="gz-sim-joint-state-publisher-system" 
            name="gz::sim::systems::JointStatePublisher">
            <joint_name>link_1_joint</joint_name>
            <joint_name>link_2_joint</joint_name>
            <joint_name>link_3_joint</joint_name>
            <update_rate>100</update_rate>
        </plugin>
    </gazebo>

</robot>
````



Tras la sintonización experimental, se establecieron los siguientes valores para garantizar el desempeño óptimo del controlador:

# **Validación de Control: Rutina de Prueba PTP**

Para verificar la estabilidad del controlador PID sintonizado y el rango de movimiento del robot, se desarrolló una rutina de prueba punto a punto (PTP).

Este script envía comandos de posición tipo "escalón" (step inputs) a las tres articulaciones simultáneamente. El objetivo es estresar el sistema de control solicitando cambios inmediatos de posición y observar el comportamiento de frenado en cada parada.

### Código de la Rutina de Prueba (`test_routine.py`)

El siguiente script en Python utiliza la librería rclpy para publicar en los tópicos de comando (/cmd_pos). Se definió una lista de waypoints estratégicos que llevan al robot a sus límites y al centro, con una pausa de 3 segundos entre movimientos para evaluar la estabilidad en estado estacionario.

````python
import rclpy
from rclpy.node import Node
from std_msgs.msg import Float64
import time

class ScaraRoutine(Node):
    def __init__(self):
        super().__init__('scara_routine_node')
        
        # 1. Creamos los publicadores para cada articulación
        # Se definen los topics correspondientes a los controladores de posición
        self.pub_j1 = self.create_publisher(Float64, '/joint1/cmd_pos', 10)
        self.pub_j2 = self.create_publisher(Float64, '/joint2/cmd_pos', 10)
        self.pub_j3 = self.create_publisher(Float64, '/joint3/cmd_pos', 10)

        self.get_logger().info('¡Iniciando rutina de prueba SCARA!')

    def move_robot(self, j1_val, j2_val, j3_val):
        """Función auxiliar para publicar en los 3 motores a la vez"""
        msg1 = Float64()
        msg1.data = float(j1_val)
        
        msg2 = Float64()
        msg2.data = float(j2_val)
        
        msg3 = Float64()
        msg3.data = float(j3_val)

        self.pub_j1.publish(msg1)
        self.pub_j2.publish(msg2)
        self.pub_j3.publish(msg3)
        
        self.get_logger().info(f'Enviando -> J1: {j1_val}, J2: {j2_val}, J3: {j3_val}')

def main(args=None):
    rclpy.init(args=args)
    node = ScaraRoutine()

    # --- LISTA DE POSICIONES A PROBAR (Waypoints) ---
    # Formato: [Junta1 (rad), Junta2 (rad), Junta3 (m)]
    waypoints = [
        [0.0,  0.0,  0.0],   # 1. HOME (Centro)
        [0.5,  0.5,  0.1],   # 2. Movimiento suave derecha
        [1.0,  1.0,  0.2],   # 3. Extensión máxima derecha
        [0.0,  0.0,  0.0],   # 4. Regreso al centro
        [-0.5, -0.5, 0.1],   # 5. Movimiento suave izquierda
        [-1.0, -1.0, 0.2],   # 6. Extensión máxima izquierda
        [1.5, -1.5,  0.0],   # 7. Cruce (J1 derecha, J2 izquierda) - Prueba colisiones
        [0.0,  0.0,  0.0]    # 8. Finalizar en Home
    ]

    try:
        # Damos un segundo para que ROS conecte los nodos
        time.sleep(1)

        for point in waypoints:
            # Extraemos los valores
            val_j1, val_j2, val_j3 = point
            
            # Movemos el robot
            node.move_robot(val_j1, val_j2, val_j3)
            
            # --- TIEMPO DE ESPERA ---
            # Pausa de 3 segundos para observar el asentamiento (settling time)
            # y verificar la ausencia de oscilaciones residuales.
            time.sleep(3.0) 

    except KeyboardInterrupt:
        print("\nRutina cancelada por el usuario.")
    finally:
        node.destroy_node()
        rclpy.shutdown()

if __name__ == '__main__':
    main()
````

# **Interpolación de Movimiento: Rutina de Prueba Suave**

Tras validar el funcionamiento básico con comandos escalón, se observó que los movimientos bruscos generaban estrés mecánico innecesario y pequeñas vibraciones residuales. Para solucionar esto, se implementó una segunda rutina de prueba que genera **perfiles de velocidad suaves**.

<div align="center">
    <img src="imagenesP1P2/2.gif" width="500" alt="Prueba Suave" />
    <br>
    <b>Figura 2:</b> Prueba de movimiento suave con interpolación polinómica.
</div>



Esta técnica es fundamental en robótica industrial para garantizar la longevidad de los actuadores y la precisión en el seguimiento de trayectorias.

### Algoritmo de Suavizado (SmoothStep)

El script `test_smooth.py` no envía un único comando de posición final. En su lugar, divide el trayecto en cientos de puntos intermedios y calcula la posición para cada instante de tiempo utilizando una función de interpolación polinómica conocida como **SmoothStep**:

$$S(t) = t^2 \cdot (3 - 2t)$$

Donde $t$ va de 0 a 1. Esta curva genera una aceleración inicial suave y una desaceleración progresiva al llegar al objetivo, eliminando los "latigazos" inerciales típicos del control "bang-bang".

### Código de la Rutina Suave (`test_smooth.py`)

El siguiente código implementa la lógica descrita, permitiendo definir no solo la posición objetivo, sino también la duración deseada del movimiento.

````python
import rclpy
from rclpy.node import Node
from std_msgs.msg import Float64
import time
import math

class ScaraSmooth(Node):
    def __init__(self):
        super().__init__('scara_smooth_node')
        
        # Publicadores
        self.pub_j1 = self.create_publisher(Float64, '/joint1/cmd_pos', 10)
        self.pub_j2 = self.create_publisher(Float64, '/joint2/cmd_pos', 10)
        self.pub_j3 = self.create_publisher(Float64, '/joint3/cmd_pos', 10)
        
        # Estado inicial del robot
        self.current_j1 = 0.0
        self.current_j2 = 0.0
        self.current_j3 = 0.0

        self.get_logger().info('¡Controlador de Trayectorias Suaves listo!')

    def publish_pos(self, j1, j2, j3):
        msg1 = Float64(); msg1.data = float(j1)
        msg2 = Float64(); msg2.data = float(j2)
        msg3 = Float64(); msg3.data = float(j3)
        
        self.pub_j1.publish(msg1)
        self.pub_j2.publish(msg2)
        self.pub_j3.publish(msg3)

    def move_smooth(self, target_j1, target_j2, target_j3, duration=3.0):
        """
        Genera una trayectoria interpolada desde la posición actual hasta la objetivo.
        """
        self.get_logger().info(f'Iniciando movimiento suave hacia: {target_j1}, {target_j2}, {target_j3}')

        # Configuración de la frecuencia de interpolación
        frequency = 50.0  # 50 Hz (envío de comandos cada 20ms)
        period = 1.0 / frequency
        steps = int(duration * frequency)

        # Puntos de inicio y diferencia
        start_j1, start_j2, start_j3 = self.current_j1, self.current_j2, self.current_j3
        diff_j1 = target_j1 - start_j1
        diff_j2 = target_j2 - start_j2
        diff_j3 = target_j3 - start_j3

        # BUCLE DE GENERACIÓN DE TRAYECTORIA
        for i in range(steps + 1):
            # 't' normalizado de 0.0 a 1.0
            t = i / float(steps)

            # --- INTERPOLACIÓN SMOOTH STEP (Curva en S) ---
            # Transforma el tiempo lineal en una curva de aceleración/desaceleración
            smooth_t = t * t * (3 - 2 * t)
            
            # Cálculo de posición intermedia
            next_j1 = start_j1 + (diff_j1 * smooth_t)
            next_j2 = start_j2 + (diff_j2 * smooth_t)
            next_j3 = start_j3 + (diff_j3 * smooth_t)

            self.publish_pos(next_j1, next_j2, next_j3)
            time.sleep(period)

        # Actualización de estado interno
        self.current_j1 = target_j1
        self.current_j2 = target_j2
        self.current_j3 = target_j3
        self.get_logger().info('Movimiento finalizado.')

def main(args=None):
    rclpy.init(args=args)
    node = ScaraSmooth()

    try:
        time.sleep(1)
        # Secuencia de movimientos
        node.move_smooth(1.0, 0.8, 0.2, duration=4.0)  # Extensión lenta
        time.sleep(1.0)
        node.move_smooth(-1.0, -0.8, 0.1, duration=3.0) # Cambio de lado
        time.sleep(1.0)
        node.move_smooth(0.0, 0.0, 0.0, duration=5.0)   # Retorno suave a Home

    except KeyboardInterrupt:
        pass
    finally:
        node.destroy_node()
        rclpy.shutdown()

if __name__ == '__main__':
    main()
````

# **Control en Espacio Cartesiano: Cinemática Inversa**

Como prueba final y demostración de control avanzado, se implementó un nodo capaz de mover el robot basándose en coordenadas cartesianas (X, Y) en lugar de ángulos articulares.

Esta capacidad es esencial para aplicaciones reales donde el robot debe seguir una trayectoria geométrica definida en el espacio de trabajo, independientemente de la configuración de sus eslabones.

    <img src="imagenesP1P2/3.gif" width="500" alt="Prueba IK" />
    <br>
    <b>Figura 3:</b> Trazado de trayectoria circular mediante Cinemática Inversa.
</div>

### Algoritmo de Cinemática Inversa (IK)

Para un robot SCARA de dos eslabones planos, la posición (x,y) del efector final se relaciona con los ángulos de las articulaciones q1​ y q2​ mediante ecuaciones trigonométricas derivadas de la geometría del robot.

El algoritmo implementado resuelve estas ecuaciones en tiempo real:

    Utiliza el teorema de los cosenos para determinar el ángulo del codo (q2​) necesario para alcanzar la distancia radial requerida.

    Calcula el ángulo de la base (q1​) ajustando la orientación del brazo hacia el objetivo.

Parámetros Geométricos (URDF):

    Longitud L1​=0.45 m

    Longitud L2​=0.45 m

>Aquí está el desglose de dónde salen los números:
L1​ (Longitud del Eslabón 1): Se encuentra en el joint que conecta el link 1 con el 2 (link_2_joint).
Línea: <origin xyz="0.45 0 0.05" rpy="0 0 0"/>
El valor X es la longitud: 0.45 metros.
L2​ (Longitud del Eslabón 2): Se encuentra en el joint que conecta el link 2 con el 3 (link_3_joint).
Línea: <origin xyz="0.45 0 -0.05" rpy="0 0 0"/>
El valor X es la longitud: 0.45 metros.
>**Sabiendo lo anterior se procede a genera el codigo para la trayectoria circular , osea que la punta del eslabon 3 dibuje un circulo.**

````python
import rclpy
from rclpy.node import Node
from std_msgs.msg import Float64
import time
import math

class ScaraIK(Node):
    def __init__(self):
        super().__init__('scara_ik_node')
        
        # Publicadores para los 3 motores
        self.pub_j1 = self.create_publisher(Float64, '/joint1/cmd_pos', 10)
        self.pub_j2 = self.create_publisher(Float64, '/joint2/cmd_pos', 10)
        self.pub_j3 = self.create_publisher(Float64, '/joint3/cmd_pos', 10)
        
        # --- DIMENSIONES REALES DEL ROBOT (Según URDF) ---
        self.L1 = 0.45  # Longitud del Link 1
        self.L2 = 0.45  # Longitud del Link 2

        self.get_logger().info(f'¡IK lista! Longitudes: L1={self.L1}, L2={self.L2}')

    def publish_joints(self, q1, q2, q3):
        """Publica los ángulos en los topics"""
        msg1 = Float64(); msg1.data = float(q1)
        msg2 = Float64(); msg2.data = float(q2)
        msg3 = Float64(); msg3.data = float(q3)
        
        self.pub_j1.publish(msg1)
        self.pub_j2.publish(msg2)
        self.pub_j3.publish(msg3)

    def inverse_kinematics(self, x, y):
        """
        Calcula la Cinemática Inversa (IK) para un brazo plano.
        Entrada: Coordenadas (x, y) deseadas.
        Salida: Ángulos (q1, q2).
        """
        try:
            # 1. Teorema de cosenos para encontrar el ángulo del codo (q2)
            dist_sq = x**2 + y**2
            
            # Ley de cosenos despejada:
            cos_q2 = (dist_sq - self.L1**2 - self.L2**2) / (2 * self.L1 * self.L2)
            
            # Verificar si el punto es alcanzable (dentro del espacio de trabajo)
            if cos_q2 > 1.0 or cos_q2 < -1.0:
                return None, None

            sin_q2 = math.sqrt(1 - cos_q2**2)
            q2 = math.atan2(sin_q2, cos_q2)

            # 2. Calcular ángulo de la base (q1)
            k1 = self.L1 + self.L2 * cos_q2
            k2 = self.L2 * sin_q2
            q1 = math.atan2(y, x) - math.atan2(k2, k1)

            return q1, q2

        except ValueError:
            return None, None

    def draw_circle(self):
        """Genera trayectoria circular y mueve el robot"""
        # Configuración del círculo
        center_x = 0.5   # Centro en X
        center_y = 0.0   # Centro en Y
        radius = 0.15    # Radio (15 cm)
        
        steps = 300      # Resolución de la trayectoria
        
        for i in range(steps):
            # Parametrización (0 a 2PI)
            theta = (i / steps) * 2 * math.pi
            
            # Calcular coordenada cartesiana objetivo
            target_x = center_x + radius * math.cos(theta)
            target_y = center_y + radius * math.sin(theta)
            
            # Calcular ángulos necesarios (IK)
            q1, q2 = self.inverse_kinematics(target_x, target_y)
            
            if q1 is not None:
                # q3 fijo (muñeca)
                self.publish_joints(q1, q2, 0.0)
            
            time.sleep(0.02)

def main(args=None):
    rclpy.init(args=args)
    node = ScaraIK()

    try:
        time.sleep(1)
        node.get_logger().info('Yendo a la posición inicial...')
        
        # 1. Aproximación al punto de inicio
        start_x = 0.5 + 0.15
        start_y = 0.0
        q1, q2 = node.inverse_kinematics(start_x, start_y)
        
        if q1 is not None:
            node.publish_joints(q1, q2, 0.0)
            time.sleep(3) # Estabilización
        
        node.get_logger().info('¡Dibujando!')
        
        # 2. Ejecución continua
        while rclpy.ok():
            node.draw_circle()
            
    except KeyboardInterrupt:
        pass
    finally:
        node.destroy_node()
        rclpy.shutdown()

if __name__ == '__main__':
    main()
````

## Conclusiones Practica 1 y 2

A lo largo de estas practicas, logramos transformar un simple modelo 3D en un sistema robótico funcional dentro de Gazebo, comprendiendo que la simulación física requiere mucho más que solo definir la geometría del robot. La adaptación del código por parte del profesor fue clave para entender cómo ROS 2 interactúa con el simulador a través de los plugins de control (gz_ros2_control), convirtiendo nuestro URDF estático en una máquina dinámica capaz de responder a comandos.

Uno de los aprendizajes más importantes fue el proceso de sintonización del PID. Fue necesario ajustar experimentalmente los valores, aumentando significativamente la ganancia derivativa (Kd​) en la base para lograr ese frenado "seco" y preciso que se busca en la industria, y equilibrando la fuerza proporcional (Kp​) para tener una respuesta rápida sin rebotes.

La implementación de las rutinas de prueba en Python nos permitió validar progresivamente el sistema:

Con (`test_smooth.py`), dimos un salto de calidad al implementar interpolación polinómica. Ver al robot moverse con aceleraciones y desaceleraciones suaves nos mostró la importancia de planificar las trayectorias para cuidar los actuadores y mejorar la estética del movimiento.

Finalmente, con (`scara_ik_circle.py`), aplicamos la teoría de Cinemática Inversa. Fue muy satisfactorio ver cómo las matemáticas (teorema de cosenos y trigonometría) coincidían perfectamente con la realidad simulada, permitiendo controlar el robot en coordenadas cartesianas (X, Y) para dibujar un círculo perfecto, lo cual abre la puerta a aplicaciones reales como pick-and-place o trazado.
